```python
import pandas as pnd
```

```python
# Downloading the benchmark file using pandas library
benchmark_file = pnd.read_csv("HGVS_2014_benchmark.tsv", sep="\t")
```

```python
# Printing benchmark file read with pandas
print(benchmark_file)
print("length of the benchmark_file:",len(benchmark_file))
```

```python
# For counting of "Pathogenic and Benign" using pandas
benign_count = 0
pathogenic_count = 0

for row in range(len(benchmark_file)):
    value = benchmark_file.iloc[row,1]

    if value == "Benign":
        benign_count += 1
    elif value == "Pathogenic":
        pathogenic_count += 1

print(f"\nBenign: {benign_count}")
print(f"Pathogenic: {pathogenic_count}")
    
```
